```python
plt.figure(figsize=(10, 6))
sns.histplot(data, x = 'Sales', bins = 40, color = 'steelblue', edgecolor = 'black')
plt.xlim(0, 3000)
plt.title('Distribution of Sales (0–3000)')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data, x = 'Sales', y = 'Category', hue = 'Category', palette = 'Set2', legend = False, width = 0.5, linewidth = 1.5, showfliers = False)
plt.title('Sales Distribution by Category', fontsize = 16, fontweight = 'bold')
plt.xlabel('Sales')
plt.ylabel('Category')
sns.despine()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(11, 6))
sns.scatterplot(data, x = 'Row ID', y = 'Sales', color = '#3498DB', alpha = 0.35, s = 30, edgecolor=None)
plt.title('Sales Across Records', fontsize=17, fontweight = 'bold', pad = 15)
plt.xlabel('Record ID', fontsize=12)
plt.ylabel('Sales', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.25)
sns.despine()
plt.tight_layout()
plt.show()
```
